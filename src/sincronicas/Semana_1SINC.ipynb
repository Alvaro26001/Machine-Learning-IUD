## Transformacion de los datos 

### Se traducen los nombres de las columnas y visualizamos los cambios

```python
nuevos_nombres = {
    "longitude": "longitud",
    "latitude": "latitud",
    "housing_median_age": "antiguedad_media_vivienda",
    "total_rooms": "total_habitaciones",
    "total_bedrooms": "total_dormitorios",
    "population": "poblacion",
    "households": "hogares",
    "median_income": "ingreso_mediano",
    "median_house_value": "valor_mediano_vivienda",
    "ocean_proximity": "proximidad_oceano"
}


df = df.rename(columns=nuevos_nombres)

df.head()
```

## Confirmamos los cambios

```python
print("nombre de las columnas:")
print(df.columns.tolist())
```

### En el siguiente codigo se remplazaran los tipos de datos de las columnas "total_habitaciones,total_dormitorios,poblacion,hogares" el tipo de datos que tenemos para estos datos son de tipo float64 y los cambairemos a tipo Int64, utilizaremos "Int64" ya que tenemos valores nulos dentro de total_dormitorios y en este paso no trasformaremos aun esos datos 

```python
columnas_enteras = [
    "total_habitaciones",
    "total_dormitorios",
    "poblacion",
    "hogares"
]

for columna in columnas_enteras:
    df.loc[:, columna] = df[columna].astype("Int64")
```

## Verificacion del cambio de tipo en los datos

```python
print(df.dtypes)
```

### En el siguiente codigo tranformaremos lo valores nulos en "total_dormitorios" primero visualizaremos y entederemos los datos que estan en esta columna para determinar que procedimiento es mejor y luego visalizaremos los cambios realizados 

```python
# Cantidad de valores nulos
print("Valores nulos:", df["total_dormitorios"].isnull().sum())

print(df["total_dormitorios"].describe())

# Calcular mediana
print("Mediana:", df["total_dormitorios"].median())
```

## Utlizaremos el valor de la mediana para rellenar estos datos 

```python
# Calcular la mediana
mediana_dormitorios = int(df["total_dormitorios"].median())

# Reemplazar valores nulos
df.loc[:, "total_dormitorios"] = (
    df["total_dormitorios"].fillna(mediana_dormitorios)
)
```

## visualizamos los cambios 

```python
# Comprobar valores nulos restantes
print("Valores nulos restantes:",
      df["total_dormitorios"].isnull().sum())

# Comprobar tipo de dato
print("Tipo de dato:",
      df["total_dormitorios"].dtype)

# Estadísticas actualizadas
print(df["total_dormitorios"].describe())
```

### En la siguiente transformacion cambiaremos nuestros datos categoricos de la columna "proximidad_oceano",para esto ustilizaremos de forma preliminar la tecnica One-Hot Encoding, pero antes traduciremos las categorias dentro de "proximidad_oceano" para mantener nuestra coherencia respecto a mantener nuestro dataFrame en español 

```python
traduccion_oceano = {
    "NEAR BAY": "cerca_bahia",
    "NEAR OCEAN": "cerca_oceano",
    "ISLAND": "isla",
    "INLAND": "interior",
    "<1H OCEAN": "una_hora_oceano"
}

df.loc[:, "proximidad_oceano"] = (
    df["proximidad_oceano"].replace(traduccion_oceano)
)
```

## Verificamos que los cambios fueron generados 

```python
print(df["proximidad_oceano"].value_counts())
```

## Aplicando  One-Hot Encoding

```python
# Transformación de variables categóricas
df = pd.get_dummies(
    df,
    columns=["proximidad_oceano"],
    dtype="int64"
)
```
